```python
import pandas as pd
import numpy as np

df = pd.read_csv("train.csv")

print("Dataset loaded successfully")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
print(df.isnull().sum())
```

Missing Data Handling Justification

Age:
Median imputation is used because Age is a numerical variable and the median is less affected by extreme values than the mean.

Embarked:
Mode imputation is used because Embarked is a categorical variable and the most frequent category is a reasonable replacement for missing values.

Cabin:
Missing Cabin values are replaced with "Unknown" instead of deleting rows because removing all rows with missing Cabin information would result in significant data loss.

```python
if "Age" in df.columns:
    df["Age"] = df["Age"].fillna(df["Age"].median())

if "Embarked" in df.columns:
    df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])

if "Cabin" in df.columns:
    df["Cabin"] = df["Cabin"].fillna("Unknown")

print("Missing values handled")
```

```python
print("Missing values after handling:")
print(df.isnull().sum())
```

```python
duplicates_before=df.duplicated().sum()
print("Duplicates before removal:", duplicates_before)
```

```python
df = df.drop_duplicates()

duplicates_after = df.duplicated().sum()

print("Duplicates removed:", duplicates_before - duplicates_after)
print("Duplicates remaining:", duplicates_after)
```

```python
for column in df.select_dtypes(include="object").columns:
    df[column] = df[column].astype(str).str.strip()
```

```python
if "Sex" in df.columns:
    df["Sex"] = df["Sex"].replace({
        "male": "Male",
        "MALE": "Male",
        "m": "Male",
        "M": "Male",
        "female": "Female",
        "FEMALE": "Female",
        "f": "Female",
        "F": "Female"
    })

print(df["Sex"].value_counts())
```

```python
if "Age" in df.columns:
    df["Age"] = pd.to_numeric(df["Age"], errors="coerce")

if "Fare" in df.columns:
    df["Fare"] = pd.to_numeric(df["Fare"], errors="coerce")

if "SibSp" in df.columns:
    df["SibSp"] = pd.to_numeric(df["SibSp"], errors="coerce")

if "Parch" in df.columns:
    df["Parch"] = pd.to_numeric(df["Parch"], errors="coerce")
```

```python
if "PassengerId" in df.columns:
    df["PassengerId"] = df["PassengerId"].astype(str)

if "Ticket" in df.columns:
    df["Ticket"] = df["Ticket"].astype(str)

print(df.dtypes)
```

```python
outlier_summary = []

numeric_columns = df.select_dtypes(include=np.number).columns

for column in numeric_columns:

    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    outlier_summary.append({
        "Column": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Limit": lower_limit,
        "Upper Limit": upper_limit,
        "Outlier Count": len(outliers)
    })

outlier_report = pd.DataFrame(outlier_summary)

print(outlier_report)
```

```python
print("Final Shape:", df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:")
print(df.duplicated().sum())

print("\nData Types:")
print(df.dtypes)
```

```python
after_rows = len(df)
after_duplicates = df.duplicated().sum()
after_nulls = df.isnull().sum().sum()

after_summary = pd.DataFrame({
    "Metric": [
        "Row Count",
        "Duplicate Rows",
        "Total Null Values"
    ],
    "Before Cleaning": [
        before_rows,
        before_duplicates,
        before_nulls
    ],
    "After Cleaning": [
        after_rows,
        after_duplicates,
        after_nulls
    ]
})

print(after_summary)
```

```python
df.to_csv("cleaned_titanic.csv", index=False)

print("Cleaned dataset saved successfully")
```
